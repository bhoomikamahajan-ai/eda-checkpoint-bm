## Visualize

Create any visualizations for your EDA here. Make note in the form of code comments of what your thought process is for your visualizations.

```python
import matplotlib.pyplot as plt


# Clean hidden characters: I had to do this clean up part as the visualization was not working because of the carriage return and tab
text_cols = df.select_dtypes(include="object").columns
df[text_cols] = df[text_cols].replace({r'[\r\n\t]': ' '}, regex=True)

# Converting Year column to numeric

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

# Counting the job posts per year
posts_per_year = df.groupby("Year")["jobpost"].count()


plt.figure(figsize=(10,5))
plt.plot(posts_per_year.index, posts_per_year.values)
plt.xlabel("Year")
plt.ylabel("Number of Job Posts")
plt.title("Job Postings Over Time")
plt.show()
```

```python
# Visualization 2: Top 10 most common job titles
# Thought process: Identify the roles most in demand.
# ------------------------------------------------------------------

top_titles = df['Title'].value_counts().head(10)

plt.figure(figsize=(10,5))
plt.bar(top_titles.index, top_titles.values, edgecolor = "black")
plt.xticks(rotation=45, ha='right')
plt.xlabel("Job Title")
plt.ylabel("Count")
plt.title("Top 10 Most Common Job Titles")
plt.show()
```
